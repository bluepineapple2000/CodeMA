```python
from PIL import Image
import os
from pathlib import Path
```

```python
def merge_images(path1, path2, output_path):
    img1 = Image.open(path1).convert("RGB")
    img2 = Image.open(path2).convert("RGB")

    # Make sure both images are the same size
    assert img1.size == img2.size, "Images must be the same size!"

    pixels1 = img1.load()
    pixels2 = img2.load()

    width, height = img1.size
    result = Image.new("RGB", (width, height))
    result_pixels = result.load()

    for y in range(height):
        for x in range(width):
            r1, g1, b1 = pixels1[x, y]
            r2, g2, b2 = pixels2[x, y]

            # Treat as white if any channel is above a threshold
            is_black1 = (r1 + g1 + b1) < 350  # roughly > 127 per channel
            is_black2 = (r2 + g2 + b2) < 350

            if is_black1 or is_black2:
                result_pixels[x, y] = (0, 0, 0)  # white
            else:
                result_pixels[x, y] = (255, 255, 255)        # black

    result.save(output_path)
    print(f"Saved merged image to {output_path}")
```

```python
# Get all jpg files from the directory and sort them
jpg_dir = "../data/jpg_122"
output_dir = "../data/combined_122"

# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Get all jpg files and sort them
jpg_files = sorted([f for f in os.listdir(jpg_dir) if f.lower().endswith('.jpg')])

print(f"Found {len(jpg_files)} jpg files")
print("Files:", jpg_files)

# Combine files in pairs
for i in range(0, len(jpg_files) - 1, 2):
    file1 = jpg_files[i]
    file2 = jpg_files[i + 1]
    
    path1 = os.path.join(jpg_dir, file1)
    path2 = os.path.join(jpg_dir, file2)
    
    # Create output filename from the two input files
    base1 = os.path.splitext(file1)[0]
    base2 = os.path.splitext(file2)[0]
    output_filename = f"combined_{base1}_{base2}.png"
    output_path = os.path.join(output_dir, output_filename)
    
    print(f"\nCombining: {file1} + {file2}")
    print(f"Output: {output_filename}")
    
    merge_images(path1, path2, output_path)
```
