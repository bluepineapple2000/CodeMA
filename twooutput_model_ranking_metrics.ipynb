# Two-output model ranking metrics

Reads `eva_server/model_comparison/twooutputs/model_ranking.csv`, selects the requested metric columns, adds a combined Dice/RMSE score, and displays the table.

```python
from pathlib import Path

import pandas as pd

csv_path = Path("eva_server/model_comparison/twooutputs/model_ranking.csv")
ranking = pd.read_csv(csv_path)

table = ranking.rename(columns={"spot_recognition_mean": "spot-reco-mean"})[
    [
        "model_display",
        "spot-reco-mean",
        "dice_mean",
        "iou_mean",
        "rmse_mean",
        "nrmse_mean",
        "soft_intensity_dice_mean",
    ]
].copy()

table["dice_rmse_score"] = 0.5 * table["dice_mean"] + 0.5 * (1 - table["rmse_mean"])

table
```
